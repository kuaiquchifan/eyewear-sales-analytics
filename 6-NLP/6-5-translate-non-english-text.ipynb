### 读取数据

```python
import pandas as pd
dir = r"customer_reviews_non_english.csv"
df = pd.read_csv(dir)
df
```

### 去掉text前后的空格，并且去重，并且导出

```python
# 复制数据，避免修改原 df
df_cleaned = df.copy()

# 去除每行前后空格（.str.strip() 会保留 NaN）
df_cleaned['text_clean'] = df_cleaned['text_clean'].str.strip()

# 删除空值、按 text_clean 去重并重置索引
df_cleaned = (
    df_cleaned
    .dropna(subset=['text_clean'])
    .drop_duplicates(subset=['text_clean'])
    .reset_index(drop=True)
)

df_cleaned.to_csv("customer_reviews_non_english_cleaned.csv", index=False, encoding="utf-8-sig")
```
