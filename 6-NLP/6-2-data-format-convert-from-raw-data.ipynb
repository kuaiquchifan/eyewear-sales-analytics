### 转换亚马逊商品信息的raw data格式

```python
import pandas as pd

product_data_path = r"Amazon-Reviews-2023\raw_meta_fashion\meta_Amazon_Fashion.jsonl"


df_product = pd.read_json(product_data_path, lines=True)

df_product.to_parquet("meta_Amazon_Fashion.parquet", engine="pyarrow", index=False, compression="snappy")
```

### 转换亚马逊商品评论的raw data格式

```python
import pandas as pd

review_data_path = r"Amazon-Reviews-2023\raw_meta_fashion\Amazon_Fashion_review.jsonl"

df_review = pd.read_json(review_data_path, lines=True)

df_review.to_parquet("Amazon_Fashion_review.parquet", engine="pyarrow", index=False, compression="snappy")
```
