```python
import pandas as pd
from sqlalchemy import create_engine

# 数据库配置
username = "XXXXXX"
password = "YYYYYY"
host = "localhost"
port = 5432
database = "eyewear-data"

# 创建连接
engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

# 查询数据
sql = """
SELECT
    cc.complaint_id,
    cc.complaint_type,
    cc.complaint_text,
    cc.complaint_severity,
    cc.resolution_status,
    cc.complaint_date,
    cc.resolution_date

FROM  "CustomerComplaint" cc

"""

df = pd.read_sql(sql, engine)

# 查看数据
df
```
