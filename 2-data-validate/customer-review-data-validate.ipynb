```python
import pandas as pd
from sqlalchemy import create_engine

# 数据库配置
username = "XXXXXX"
password = "YYYYYY"
host = "localhost"
port = 5432
database = "eyewear-data"

# 创建连接
engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

# 查询数据
sql = """
SELECT
    cr.review_id,
    cr.rating,
    cr.review_title,
    cr.review_text,
    cr.review_date
FROM "CustomerReview" cr;
"""

df = pd.read_sql(sql, engine)

# 查看数据
df
```
