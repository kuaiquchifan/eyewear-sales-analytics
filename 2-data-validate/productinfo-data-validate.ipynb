### Connect postgresql database

```python
import pandas as pd
from sqlalchemy import create_engine

# 数据库配置
username = "XXXXXX"
password = "YYYYYY"
host = "localhost"
port = 5432
database = "eyewear-data"

# 创建连接
engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

# 查询数据
sql = """
SELECT *
FROM "ProductInfo";
"""

df = pd.read_sql(sql, engine)

# 查看数据
df
```

```python
df.columns
```

```python
# 查看亚太区的数据
asia_pacific_df = df[df['region']=='Asia-Pacific']

# asia_pacific_df['country'].unique()
asia_pacific_df['country'].drop_duplicates().to_list()
```

```python
# 查看 EMEA 区的数据
emea_df = df[df['region']=='EMEA']
emea_df['country'].drop_duplicates().to_list()
```

```python

north_america_df = df[df['region']=='North America']
north_america_df['country'].drop_duplicates().to_list()
```

```python

latin_america_df = df[df['region']=='Latin America']
latin_america_df['country'].drop_duplicates().to_list()
```

```python
# 关闭数据库连接
engine.dispose()
```
