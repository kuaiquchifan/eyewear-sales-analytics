```python
import pandas as pd
from sqlalchemy import create_engine

# 数据库配置
username = "XXXXXX"
password = "YYYYYY"
host = "localhost"
port = 5432
database = "eyewear-data"

# 创建连接
engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

# 查询数据
sql = """
SELECT
    ci.customer_id,
    ci.customer_type,
    ci.customer_hierarchy,

    cr.review_id,
    cr.rating,
    cr.review_title,
    cr.review_text,
    cr.review_date,

    pi.product_id,
    pi.product_name,
    pi.category,

    cc.complaint_id,
    cc.complaint_type,
    cc.complaint_text,
    cc.complaint_severity,
    cc.resolution_status,
    cc.complaint_date,
    cc.resolution_date

FROM "CustomerInfo" ci

LEFT JOIN "CustomerReview" cr
    ON ci.customer_id = cr.customer_id

LEFT JOIN "CustomerComplaint" cc
    ON ci.customer_id = cc.customer_id

LEFT JOIN "ProductInfo" pi
    ON cr.product_id = pi.product_id;
"""

df = pd.read_sql(sql, engine)

# 查看数据
df
```
