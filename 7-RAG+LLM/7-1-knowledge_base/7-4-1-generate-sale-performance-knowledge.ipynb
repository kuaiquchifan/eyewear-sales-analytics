### 输出为 Jsonl 格式

```python
import json
from pathlib import Path

# 1) 目标目录
knowledge_dir = Path(r".\7-RAG+LLM\output")
knowledge_dir.mkdir(parents=True, exist_ok=True)

# 2) 数据表映射
df_map = {
    "monthly": df_monthly,
    "quarterly": df_quarterly,
    "yearly": df_yearly,
}

# 3) 统一字段生成：按“时间粒度 + 业务指标”生成知识内容
def build_knowledge_content(row, time_unit):
    # 兼容不同 DataFrame 的列名
    keys = row.to_dict()
    
    # 统一读取常见字段
    year = keys.get("year")
    month = keys.get("month")
    quarter = keys.get("quarter")
    brand = keys.get("brand") or "all brands"
    product = keys.get("product_name") or keys.get("product") or keys.get("category") or "all products"
    country = keys.get("country") or keys.get("country_name") or "USA"
    channel = keys.get("channel") or "offline"
    total_qty = keys.get("total_qty") or keys.get("quantity") or 0
    total_revenue = keys.get("total_revenue") or keys.get("sales_amount") or keys.get("sales") or 0
    gross_margin = keys.get("gross_margin") or keys.get("gross_profit") or 0
    avg_order_value = keys.get("avg_order_value") or keys.get("AOV") or 0

    time_label = ""
    if time_unit == "monthly":
        time_label = f"{year}-{month:02d}" if pd.notna(year) and pd.notna(month) else str(year or month or "")
    elif time_unit == "quarterly":
        time_label = f"{year}-Q{quarter}" if pd.notna(year) and pd.notna(quarter) else str(year or quarter or "")
    elif time_unit == "yearly":
        time_label = str(year) if pd.notna(year) else ""

    # 业务描述文本
    desc = (
        f"{time_label} 的{time_unit}销售表现为："
        f"销售数量 {total_qty}，销售额 {total_revenue}，"
        f"毛利 {gross_margin}，平均订单价值 {avg_order_value}。"
    )
    if brand:
        desc += f"品牌维度为 {brand}。"
    if product:
        desc += f"产品/品类维度为 {product}。"
    if country:
        desc += f"覆盖地区为 {country}。"
    if channel:
        desc += f"销售渠道为 {channel}。"

    return desc

# 4) 导出 JSONL
for time_unit, df in df_map.items():
    df_out = df.copy()
    
    # 确保行索引不丢
    df_out = df_out.reset_index(drop=True)
    
    # 统一增加字段，适合 RAG
    df_out["time_unit"] = time_unit
    df_out["content"] = df_out.apply(lambda row: build_knowledge_content(row, time_unit), axis=1)
    
    # 让每条记录都保留原始字段 + content
    records = df_out.to_dict(orient="records")
    
    out_path = knowledge_dir / f"7-4-1-{time_unit}_sales_performance_knowledge.jsonl"
    with open(out_path, "w", encoding="utf-8-sig", newline="") as f:
        for row in records:
            f.write(json.dumps(row, ensure_ascii=False) + "\n")
    
    print(f"已导出: {out_path}")
    print(f"记录数: {len(records)}")
```
