```python
import pandas as pd
from sqlalchemy import create_engine, text

# 数据库配置
username = "XXXXXX"
password = "YYYYYY"
host = "localhost"
port = 5432
database = "eyewear-data"

# 创建连接
engine = create_engine(
    f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}"
)

# 查询数据
sql = """
SELECT *
FROM "Order";
"""

df = pd.read_sql(sql, engine)

# 查看数据
df
```

```python
df['order_status'].unique()
```

```python
sql_2 = """
SELECT order_status, COUNT(*) AS cnt
FROM "Order"
GROUP BY order_status
ORDER BY cnt DESC;
"""
df_2 = pd.read_sql(sql_2, engine)

# 查看聚合结果
df_2.to_parquet("2023_2024_order_status_aggregate.parquet",engine = 'fastparquet',index=False)
print("聚合结果已保存为 2023_2024_order_status_aggregate.parquet")
df_2
```
